<h3 style="font-size:18px;font-family: Arial; color: blue; font-weight: bold;">
Filtrado con condiciones
</h3>

```python
# 18. Mostrar los clientes con "antiguedad_laboral_anios" es igual a 20.
df[df["antiguedad_laboral_anios"]==20]
```

```python
# 19. Mostrar los clientes con "edad" mayor a 62 años.
df[df["edad"]>62]
```

```python
# 20. Mostrar los clientes con "atrasos_ultimos_12_meses" igual a 2 o 4.
df[(df["atrasos_ultimos_12_meses"]==2) | (df["atrasos_ultimos_12_meses"]==4)]
```

```python
# 21. Mostrar los clientes con "riesgo" "Alto" y "atrasos_ultimos_12_meses" mayor a 2.
df[(df["riesgo"]=="Alto") & (df["atrasos_ultimos_12_meses"]>2)]
```

```python
# 22. Mostrar los clientes cuyo "ingreso_mensual" sea menor a 1500.
df[df["ingreso_mensual"]<1500]
```

```python
# 23. Mostrar los clientes con más de 19000 en "monto_solicitado".
df[df["monto_solicitado"]>19000]
```

<h3 style="font-size:18px;font-family: Arial; color: blue; font-weight: bold;">
Creación de Tablas Resumen / Creación de Tablas de Frecuencias
</h3>

```python
# 24. Mostrar una tabla de frecuencias del "riesgo" de los clientes.
df["riesgo"].value_counts()
```

```python
# 25. Mostrar una tabla de frecuencias del "plazo_meses" de los clientes.
df["plazo_meses"].value_counts()
```

```python
# 26. Mostrar una tabla resumen del promedio del "ingreso_mensual", del "monto_solicitado" y de la "deuda_mensual" agrupado por "riesgo".
df.groupby("riesgo")[["ingreso_mensual","monto_solicitado","deuda_mensual"]].mean()
```

```python
# 27. Mostrar una tabla resumen del máximo "monto_solicitado" y "ingreso_mensual" agrupado por "riesgo".
df.groupby("riesgo")[["monto_solicitado","ingreso_mensual"]].max()
```

```python
# 28. Mostrar una tabla cruzada de frecuencias entre "riesgo" y "plazo_meses".
pd.crosstab(df["riesgo"],df["plazo_meses"])
```

```python
# 29. Mostrar una tabla cruzada de frecuencias entre "riesgo" y "atrasos_ultimos_12_meses".
pd.crosstab(df["riesgo"],df["atrasos_ultimos_12_meses"])
```
